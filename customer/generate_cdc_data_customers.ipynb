### **Generate and Load customer data**

```python
def generate_cdc_data_customers(num_rows=20):
    operations = ["INSERT", "UPDATE", "DELETE"]
    data = []
    for i in range(num_rows):
        customer_id = i + 1
        operation = random.choice(operations) # Randomly pick an operation
        ts = datetime.now() - timedelta(minutes=random.randint(1, 100))

        record = {
            "customer_id": customer_id,
            "name": f"Customer_{customer_id}",
            "email": f"customer{customer_id}@example.com",
            "phone": f"+47-943-{random.randint(1000, 9999)}",
            "operation": operation,
            "time_stamp": ts.strftime("%Y-%m-%d %H:%M:%S")
        }
        data.append(record)
    return pd.DataFrame(data) 
    
```
